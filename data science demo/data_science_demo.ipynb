# Data Science Demo

This notebook loads some dummy data into a pandas DataFrame and demonstrates the main dataframe operations you can use in a quick demo.

```python
%pip install pandas
```

```python
import pandas as pd

# Create dummy sales data
data = {
    'customer_id': [101, 102, 103, 104, 105, 106, 107, 108],
    'region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone'],
    'sales': [1200, 450, 700, 980, 600, 820, 1100, 520],
    'units_sold': [15, 30, 20, 12, 25, 18, 22, 28],
    'discount': [0.05, 0.10, 0.08, 0.04, 0.12, 0.07, 0.09, 0.06]
}

df = pd.DataFrame(data)
df.head(5)
```

## Common dataframe operations

- View the data
- Select columns
- Filter rows
- Sort values
- Group and aggregate
- Summarize statistics

```python
# Select a few columns
df[['customer_id', 'region', 'sales']]

# Filter rows
filtered_df = df[df['sales'] > 800]
filtered_df

# Sort by sales descending
df.sort_values('sales', ascending=False)

# Group by region and calculate total sales
df.groupby('region')['sales'].sum()

# Create a new calculated column
df['revenue_per_unit'] = df['sales'] / df['units_sold']
df.head()

# Basic summary statistics
df.describe()
```
